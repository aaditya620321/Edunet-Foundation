# <center> Seaborn

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
datasets = sns.get_dataset_names()
print(datasets)
total_datasets = len(datasets)
print("\nTotal Number of Datasets Available in Seaborn :",total_datasets)
```

**1. Scatter Plot:**

```python
df = sns.load_dataset("tips")
df
```

```python
#pd.set_option('display.max_rows',None)
#df
```

```python
df.head()
```

```python
df.info()
```

```python
plt.figure(figsize = (10,6))
sns.scatterplot(
    data = df,
    x = "total_bill",
    y = "tip",
    hue = "smoker", # different colours for smoker and non-smoker customers
    style = "sex", # different marker shapes for male and female customers
    size = "size",
    # sizes = (60,300),
    alpha = 0.7,
    edgecolor = "black",
    linewidth = 0.6,
    palette = "Greens_r"
)
plt.legend(title = "Customer Details", bbox_to_anchor = (1.02,1.02),loc = "upper left")
plt.show()
```

**2. Line Plot:**

```python
plt.figure(figsize = (10,6))
sns.lineplot(
    data = df,
    x = "size",
    y = "tip",
    hue = "time",
    errorbar = None,
    style = "time",
    markers = True,
    palette = "dark",
    markersize = 10,
    linewidth = 3
)
plt.title(
    "Average Tip by Table Size",fontsize = 17,fontweight = "bold",pad = 15
)
plt.xlabel(
    "Number of People",fontsize = 12, fontweight = "bold"
)
plt.ylabel(
    "Average Tip", fontsize = 12, fontweight = "bold"
)
plt.legend(
        title = "Meal Time"
    )
plt.xticks([1,2,3,4,5,6])
plt.show()
```

**3. Bar Plot:**

```python
plt.figure(figsize = (11,8))
ax = sns.barplot(
    data = df,
    x = "day",
    y = "total_bill",
    hue = "time",
    errorbar = None,
    palette = "bright",
    edgecolor = "black",
    linewidth = 1.2,
    width = 0.75
)

for container in ax.containers:
    ax.bar_label(
        container,fmt = "%.2f", padding  = 3 , fontsize = 9
    )
plt.title("Average Total Bill By Day",fontsize = 17, fontweight = "bold",pad = 15)
plt.xlabel("Day",fontsize = 12,fontweight = "bold")
plt.ylabel("Average Total Bill",fontsize = 12,fontweight = "bold")

plt.grid(
    axis = "y", linestyle = "--", alpha = 0.4
)
plt.legend(title = "Meal Time")
plt.tight_layout()
plt.show()
```

**4. Count plot:**

A Count Plot displays the Number of records in each category. It performs counting automatically, so no Y-axis column is required.

```python
plt.figure(figsize = (10,6))
ax = sns.countplot(
    data = df,
    x = "day",
    hue = "sex",
    palette = "Set2"
)
for container in ax.containers:
    ax.bar_label(
        container,fmt = "%.f", padding  = 3 , fontsize = 9
    )
plt.title("Number of Customer by Day",fontsize = 17, fontweight = "bold",pad = 15)
plt.xlabel("Day",fontsize = 12,fontweight = "bold")
plt.ylabel("Number of Customers",fontsize = 12,fontweight = "bold")

plt.grid(axis = "y",linestyle = "--",alpha = 0.4)
plt.legend(title = "Gender")
plt.tight_layout()
plt.show()
```

**5. Histogram:**

```python
import numpy as np
```

```python
plt.figure(figsize = (10,6))
bins = np.arange(
    df["total_bill"].min(),
    df["total_bill"].max()+3,3
)
sns.histplot(
    data = df,
    x = "total_bill",hue = "sex",bins = 15,multiple = "stack",palette = "Set1"
)
# plt.xlim(df["total_bill"].min(),df["total_bill"].max())
# plt.xticks(
#     np.arange(
#         df["total_bill"].min(),df["total_bill"].max()+1), rotation = 90
#     )
plt.xticks(bins,rotation = 90)
plt.show()
```

**6. Rug Plot:**

A Rug plot puts a small marks for every observations. We use this rug plot alongside with the Histograms/KDE Plot.

```python
sns.kdeplot( # kde : Kernel Density Estimate
    data = df,
    x = "total_bill",
    fill = True
)
sns.rugplot(
    data = df,x = "total_bill"
)
plt.show()
```

**7. Pointplot():**

A Point plot is used to show the average of a numerical variable across different categories, along with the error bars.

```python
sns.pointplot(
    data = df,
    x = 'day',
    y = 'tip',
    estimator = "mean", # other valules are: sum,min,max,std,var, mean, median
    hue = 'time',
    errorbar = None,
    palette = "Set1",
    linestyle = '--',
   dodge = True
)
```

**8. Pair Plot():**

A Pair Plot Shows the Scatter plot for every combinations of numeric columns, and histograms (or KDE ) on the Diagonal. It helps to Visually explore the relationships between the Multiple features. It is best for exploring the relationships b/w the multiple numerical columns.

```python
sns.pairplot(
    data = df,
    diag_kind = "kde",
    hue = 'sex',
    height = 1.4,
    palette = 'colorblind'
)
```

**9. Violin Plot:**

A Violin Plot combines a Box Plot with a Density Plot showing the distribution shape of the data. The Width of the Violin represents the frequency of the values. It is best for seeing both the spread and the shape of the data.

```python
sns.violinplot(
    data = df,
    x = "day",
    y = "tip",
    inner = "quartile", # others are : "box", "stick" , None
    hue = "sex",
    order = ["Sun","Thur","Sat","Fri"],
    split = True
)
plt.show()
```

**10. Boxplot():**

```python
sns.boxplot(
    data = df,
    x = "day",
    y = "tip"
)
```

**11. Heatmap():**

A Heatmap represents numerical values using colors. It helps to analyze the correlation b/w multiple numerical variables. With the help of this graph we will be able to analyze which variables are positively correlated, negatively correlated or zero correlated/no relationship.

```python
corr = df.corr(numeric_only = True) # it will calculate the correlation b/w multiple numeric variables.
corr
```

```python
sns.heatmap(
    corr , annot = True,cmap = "magma")
```

- +1: Perfect Positive Relationship
- +0.7 : Strong Positive
- +0.3 : Moderate Positive
- 0 : No Relationship
- -0.3 : Moderate Negative
- -0.7 : Strong Negative
- -1 : Perfect Negative
