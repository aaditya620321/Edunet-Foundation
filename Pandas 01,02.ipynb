**What is Pandas ?**

Pandas is a python library that is used to work with the datasets. Datasets can be any format like csv,excel , xml, sql, text, json, etc.

It has functions for analyzing, cleaning, exploring and manipulating the data.

The term "Pandas" has a reference to both "Panel Data" and "Python Data Analysis" and was developed by Wes Mckinnet in 2008.

```python
import pandas as pd
```

**Data Structures in Pandas:**

In Pandas there are 2-types of Data structures: 

- Series
- DataFrame

### Series:

A Pandas series is like a Column in a Table. It is a One-dimensional array holding data of any type.

```python
a = [1,2,"3",True, "India", 3.143]
myvar1 = pd.Series(a)
myvar1
```

**Labels:-**

If nothing else is Specified, the values are labeled with their default index number,First value has index 0, and it goes on.

```python
print(myvar1[0:4:2]) #slicing
```

**Creating a Label:-**

With the index argument, you can name your own labels.

```python
a = [1,2,3,4]
myvar2 = pd.Series(a)
myvar2
```

```python
a[2]
```

```python
myvar2 = pd.Series(a, index = ["W","X","Y","Z"])
myvar2
```

```python
index = [f"X{i}" for i in range(1,len(a)+1)]
myvar2 = pd.Series(a, index = index)
myvar2
```

```python
type(myvar2)
```

### DataFrame:

Datasets in Pandas are usually multi-dimensional tables, called as DataFrames. DataFrame is just like a Combination of multiple Series. A Series is Just like a Single Column but the DataFrame is the complete Table.

```python
mydataset = {
    'Cars' : ["BMW","Audi","Volvo","Ford"] ,
    'Year' : [1978, 1975, 1983,1965]
}
```

```python
myvar = pd.DataFrame(mydataset)
myvar
```

**iloc:** Indexed based location Fetching. It's Syntax is .iloc[row,column]

```python
myvar.iloc[2]
```

```python
myvar.iloc[2,0] # better way
```

```python
myvar.iloc[2]["Cars"]
```

**Customizing the Labels/Indexes:**

```python
myvar3 = pd.DataFrame(mydataset, index = ["A","B","C","D"])
myvar3
```

**loc:** Labelled Based Value Fetching.

```python
myvar3.loc["C"]
```

```python
myvar3.loc["C","Cars"]
```

```python
type(myvar)
```
