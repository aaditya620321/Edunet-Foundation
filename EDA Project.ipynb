```python
import pandas as pd
df = pd.read_excel('customer.xlsx')
pd.set_option('display.max_rows',None)
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df["CustomerID"] = df["CustomerID"].astype(str).str.extract(r'(\d+)')
df["CustomerID"] = "C" + df["CustomerID"].str.zfill(3)
df["CustomerID"]
```

```python
df
```

```python
df["Name"] = df["Name"].astype(str).str.strip()
df["Name"] = df["Name"].str.title()
df["Name"]
```

```python
df["Age"] = df["Age"].astype(str).str.strip()
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors = "coerce")
df["Age"]
```

```python
df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()
```

```python
df["Gender"]
```

```python
df["Gender"] = df["Gender"].replace({
    "m" : "Male",
    "f" : "Female",
    "male" :"Male",
    "female" : "Female"
})
df["Gender"]
```

```python
df["Purchase_Amount"] = df["Purchase_Amount"].astype(str).str.strip()
```

```python
df["Purchase_Amount"]
```

```python
df["Purchase_Amount"] = pd.to_numeric(df["Purchase_Amount"] , errors = "coerce")
df["Purchase_Amount"]
```

```python
df["JoinDate"] = pd.to_datetime(
    df["JoinDate"], errors = "coerce",dayfirst = True
)
```

```python
df["JoinDate"]
```

```python
df["JoinDate"].mode()
```

```python
df["Feedback"] = df["Feedback"].astype(str).str.strip().str.lower()
```

```python
df["Feedback"]
```

```python
df["Feedback"].unique()
```

```python
df["Feedback"] = df["Feedback"].replace({
    "good":"Good",
    "excellent":"Excellent",
    "avg":"Average",
    "average":"Average",
    "nan":pd.NA,
    "":pd.NA
})
```

```python
df["Feedback"] = df["Feedback"].str.title()
df["Feedback"]
```

```python
df.shape
```

```python
df = df.drop_duplicates()
df.shape
```

```python
df.isna().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Age"]
```

```python
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(df["Purchase_Amount"].median())
```

```python
df["Purchase_Amount"]
```

```python
df["JoinDate"] = df["JoinDate"].fillna(df["JoinDate"].mode()[0])
df["JoinDate"]
```

```python
df["Name"] = df["Name"].fillna("Unknown")
```

```python
df["Name"]
```

```python
df["Feedback"] = df["Feedback"].fillna("No Feedback")
```

```python
df["Feedback"]
```

```python
df.isna().sum()
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**Q.1) Show How many male Vs Female Customers you have after data cleaning.**

```python
sns.histplot(
    data = df,
    x = "Gender"
)
plt.title("Gender Distribution")
plt.show()
```

**Q.2) Find Number of Customers from each city and plot a chart for it.**

```python
df['City'].value_counts()
```

```python
plt.pie(df['City'].value_counts(),labels = df['City'].value_counts().index, autopct = '%.0f')
plt.show()
```

**Q.3) Plot the Age Distribution to see the spread of the age.**

```python
plt.hist(df['Age'],bins = 10, edgecolor = 'w')
plt.show()
```

**Q.4) Visualize the Average purchase value in each city.**

```python
avg_city = df.groupby("City")["Purchase_Amount"].mean().sort_values()
avg_city
```

```python
plt.bar(avg_city.index,avg_city.values)
plt.title("Average Purchase Amount by City")
plt.xlabel("City")
plt.ylabel("Avg. Purchase Amount")
plt.show()
```

**Q.5) Show how many Customers gave "Good", "Excellent","Average".**

```python
feedback_count = df["Feedback"].value_counts()
feedback_count
```

```python
sns.barplot(x = feedback_count.index, y = feedback_count )
```
