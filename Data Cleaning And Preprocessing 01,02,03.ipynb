**30. groupby():**

Groupby is used to split the data into Groups, apply a calculation and then combine the result.

**Syntax:**

df.groupby('column_name')['Value_Column'].aggregated_function()

**Q.1) What is the Average Calories Burned For each workout duration ?**

```python
df.groupby('Duration')['Calories'].mean()
```

**Q.2) What are the total calories Burned on each data ?**

```python
df.groupby('Date')['Calories'].sum()
```

**Q.3) What is the maximum heart rate for each workout duration ?**

```python
df.groupby('Duration')['Max_pulse'].max()
```
